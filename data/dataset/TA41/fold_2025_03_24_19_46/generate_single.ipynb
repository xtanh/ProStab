```python
import pandas as pd

# 蛋白质序列
protein_seq = "MSSMEAVFAGYQTRLEALKASDNPFAAGIAWVAGELTPLADARMPLMDQGFLRSDLTYDVPAIWDGRFFRLHDHLDRMERSCEKLRLRCPLGREELCGTLVGMAAKSGIRDGFVFLAVTRGLKQVRGNRPEDLTNNLYMFISPYVWVVSPEMQRTGAPAIVTRTVRRTPPGAFDPTVKNLQWGDFTRGLMEAQDRGAIYPILTDGDGNLTEGAGYNVVLVKDGVLYTPERGVLEGITRKTVMEIAAARGIDVRMEHVPVDLAYRADELFFCTTAAGIMPVTTLDGVRVGGGQVGPITRQIWEDYWDLHYDPKLTIEVPYR"

# 所有氨基酸类型
amino_acids = "ACDEFGHIKLMNPQRSTVWY"

# 创建空列表存储突变
mutations = []
index = 0

# 遍历序列中的每个位置
for pos, aa in enumerate(protein_seq, 1):  # 1-based索引
    # 对于每个位置，尝试突变为其他19种氨基酸
    for new_aa in amino_acids:
        if new_aa != aa:  # 只考虑与原始氨基酸不同的突变
            mutation = f"{aa}{pos}{new_aa}"
            mutations.append({
                "PDB": "TA41A",
                "MUT": mutation,
                "DDG": 0
            })
            index += 1

# 创建DataFrame
df = pd.DataFrame(mutations)
df.insert(0, "", range(len(df)))  # 添加索引列

# 保存为CSV
df.to_csv("TA41_mutations.csv", index=False)

# 显示前20行作为示例
print(df.head(20))
print(f"总共生成了{len(df)}个突变")
```
