```python
'''
Thanks for SPURS team for providing the code and data.
https://github.com/luo-group/SPURS

'''

from prostab.inference import  parse_pdb, get_prostab


model, cfg = get_prostab('/home/xy_th/PROSTAB/model_weight')


pdb_name = 'example'
pdb_path = '/home/xy_th/PROSTAB/data/inference_example/1A0N.pdb'
chain = 'A'


# 指定突变进行预测
mutation = "V1A"  
pdb_mut = parse_pdb(pdb_path, pdb_name, chain, cfg, mutation=mutation)
result_mutant = model(pdb_mut, return_logist=True)
print(f"突变 {mutation} 的得分: {result_mutant.item()}")
```
