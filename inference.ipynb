# ProStab prediction

Install the package and prepare weights as described in README. Start Jupyter from the repository root. Each mutation is predicted independently. Negative ΔΔG predicts increased stability.


```python
from pathlib import Path
from prostab import Predictor

root = Path.cwd()
predictor = Predictor(root / "model_weight", device="auto")
pdb_path = root / "data/inference_example/1A0N.pdb"
```

```python
predictor.predict(pdb_path, "A", "V1A")
```

```python
predictor.predict_many(pdb_path, "A", ["V1A", "P57A"])
```
